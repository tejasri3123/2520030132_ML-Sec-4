```python
a = [10, 5, 25, 8, 15]

max_val = a[0]
min_val = a[0]

for i in a:
    if i > max_val:
        max_val = i
    if i < min_val:
        min_val = i

print("Maximum =", max_val)
print("Minimum =", min_val)
```
