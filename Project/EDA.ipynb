```python
# ============================================================
# ACCIDENT HOTSPOT PREDICTION SYSTEM
# COMPLETE EDA - GOOGLE COLAB
# ============================================================
#
# 36 GRAPHS ARE:
#
# 01 Accident Severity Distribution
# 02 Top States
# 03 Top Cities
# 04 Accidents by Year
# 05 Accidents by Month
# 06 Accidents by Day of Week
# 07 Accidents by Hour
# 08 Vehicle Type Distribution
# 09 Number of Vehicles Involved
# 10 Casualties Distribution
# 11 Fatalities Distribution
# 12 Weather Conditions
# 13 Road Type
# 14 Road Condition
# 15 Lighting Conditions
# 16 Traffic Control
# 17 Speed Limit Distribution
# 18 Driver Age Distribution
# 19 Driver Gender
# 20 Driver License Status
# 21 Alcohol Involvement
# 22 Accident Location Details
# 23 Severity by Weather
# 24 Severity by Road Condition
# 25 Severity by Vehicle Type
# 26 Severity by Lighting
# 27 Severity by Alcohol
# 28 Severity by Year
# 29 Severity by Month
# 30 Severity by Day of Week
# 31 Severity by Road Type
# 32 Severity by Driver Gender
# 33 Severity by License Status
# 34 Casualties by Severity
# 35 Fatalities by Severity
# 36 Numeric Correlation Heatmap
#
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import shutil
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files

warnings.filterwarnings("ignore")

print("Libraries loaded successfully.")


# ============================================================
# 2. UPLOAD DATASET
# ============================================================

print("/content/accident_prediction_india.csv")

uploaded = files.upload()

csv_files = [
    filename
    for filename in uploaded.keys()
    if filename.lower().endswith(".csv")
]

if not csv_files:
    raise FileNotFoundError(
        "No CSV file was uploaded."
    )

uploaded_csv = csv_files[0]

DATA_FILE = os.path.join(
    "/content",
    uploaded_csv
)

print("\nDataset selected:")
print(DATA_FILE)


# ============================================================
# 3. LOAD DATASET
# ============================================================

df = pd.read_csv(
    DATA_FILE
)

print("\n============================================================")
print("DATASET LOADED")
print("============================================================")

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

print("\nColumns:")

for column in df.columns:
    print(" -", column)


# ============================================================
# 4. CREATE EDA FOLDERS
# ============================================================

EDA_DIR = "/content/eda"
GRAPHS_DIR = "/content/eda/graphs"
REPORTS_DIR = "/content/eda/reports"
SUMMARY_DIR = "/content/eda/summary"

for folder in [
    EDA_DIR,
    GRAPHS_DIR,
    REPORTS_DIR,
    SUMMARY_DIR
]:
    os.makedirs(
        folder,
        exist_ok=True
    )


# ============================================================
# 5. DATA QUALITY
# ============================================================

original_rows = len(df)

duplicate_count = int(
    df.duplicated().sum()
)

missing_total = int(
    df.isna().sum().sum()
)

print("\n============================================================")
print("DATA QUALITY")
print("============================================================")

print("Rows                 :", original_rows)
print("Duplicate rows       :", duplicate_count)
print("Total missing values :", missing_total)


# ============================================================
# 6. DATASET INFORMATION
# ============================================================

dataset_information = pd.DataFrame({
    "Column": df.columns,

    "Data Type": [
        str(df[column].dtype)
        for column in df.columns
    ],

    "Missing Values": [
        int(df[column].isna().sum())
        for column in df.columns
    ],

    "Missing Percentage": [
        round(
            df[column].isna().mean() * 100,
            2
        )
        for column in df.columns
    ],

    "Unique Values": [
        int(
            df[column].nunique(
                dropna=False
            )
        )
        for column in df.columns
    ]
})

dataset_information.to_csv(
    os.path.join(
        SUMMARY_DIR,
        "dataset_information.csv"
    ),
    index=False
)


# ============================================================
# 7. DESCRIPTIVE STATISTICS
# ============================================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

if numeric_columns:

    descriptive_statistics = (
        df[numeric_columns]
        .describe()
        .transpose()
    )

    descriptive_statistics.to_csv(
        os.path.join(
            SUMMARY_DIR,
            "descriptive_statistics.csv"
        )
    )


# ============================================================
# 8. MISSING VALUE REPORT
# ============================================================

missing_report = pd.DataFrame({
    "Column": df.columns,

    "Missing Values": [
        int(
            df[column].isna().sum()
        )
        for column in df.columns
    ]
})

missing_report["Missing Percentage"] = (
    missing_report["Missing Values"]
    / len(df)
    * 100
)

missing_report = missing_report.sort_values(
    "Missing Values",
    ascending=False
)

missing_report.to_csv(
    os.path.join(
        SUMMARY_DIR,
        "missing_values.csv"
    ),
    index=False
)


# ============================================================
# 9. DUPLICATE REPORT
# ============================================================

duplicate_summary = pd.DataFrame({
    "Measure": [
        "Original Rows",
        "Duplicate Rows",
        "Rows After Removing Duplicates"
    ],

    "Value": [
        original_rows,
        duplicate_count,
        original_rows - duplicate_count
    ]
})

duplicate_summary.to_csv(
    os.path.join(
        SUMMARY_DIR,
        "duplicate_summary.csv"
    ),
    index=False
)


# ============================================================
# 10. GRAPH HELPER
# SAVE + DISPLAY
# ============================================================

# ============================================================
# GRAPH HELPER - SAVE + DISPLAY INSIDE COLAB
# ============================================================

from IPython.display import display, Image

def save_graph(filename):

    path = os.path.join(
        GRAPHS_DIR,
        filename
    )

    # Save the graph first
    plt.tight_layout()

    plt.savefig(
        path,
        dpi=200,
        bbox_inches="tight"
    )

    # Close matplotlib figure
    plt.close()

    # Explicitly display the saved image in Colab
    display(
        Image(
            filename=path,
            width=800
        )
    )

    print("Saved and displayed:", filename)


# ============================================================
# 11. HELPER FOR BAR CHARTS
# ============================================================

def create_bar_chart(
    series,
    title,
    xlabel,
    ylabel,
    filename,
    rotation=35,
    limit=None
):

    values = (
        series
        .fillna("Missing")
        .astype(str)
        .value_counts()
    )

    if limit is not None:
        values = values.head(limit)

    plt.figure(
        figsize=(10, 6)
    )

    values.plot(
        kind="bar"
    )

    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)

    plt.xticks(
        rotation=rotation,
        ha="right"
    )

    save_graph(filename)


# ============================================================
# 12. HELPER FOR SEVERITY CROSS-TABS
# ============================================================

def create_severity_chart(
    category_column,
    title,
    xlabel,
    filename,
    limit=12,
    rotation=35
):

    table = pd.crosstab(
        df[category_column]
        .fillna("Missing")
        .astype(str),

        df["Accident Severity"]
        .fillna("Missing")
        .astype(str)
    )

    # Keep the most frequent categories
    if limit is not None:

        table = table.loc[
            table.sum(axis=1)
            .sort_values(
                ascending=False
            )
            .head(limit)
            .index
        ]

    plt.figure(
        figsize=(11, 6)
    )

    table.plot(
        kind="bar",
        ax=plt.gca()
    )

    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel("Number of Accidents")

    plt.xticks(
        rotation=rotation,
        ha="right"
    )

    plt.legend(
        title="Severity"
    )

    save_graph(filename)


# ============================================================
# 13. GRAPH 01
# ACCIDENT SEVERITY
# ============================================================

severity_counts = (
    df["Accident Severity"]
    .fillna("Missing")
    .astype(str)
    .value_counts()
)

plt.figure(
    figsize=(8, 5)
)

severity_counts.plot(
    kind="bar"
)

plt.title(
    "Accident Severity Distribution"
)

plt.xlabel(
    "Accident Severity"
)

plt.ylabel(
    "Number of Accidents"
)

plt.xticks(
    rotation=0
)

save_graph(
    "01_accident_severity_distribution.png"
)

severity_summary = pd.DataFrame({
    "Severity": severity_counts.index,
    "Accident Count": severity_counts.values
})

severity_summary["Percentage"] = (
    severity_summary["Accident Count"]
    / severity_summary["Accident Count"].sum()
    * 100
)

severity_summary.to_csv(
    os.path.join(
        SUMMARY_DIR,
        "severity_summary.csv"
    ),
    index=False
)


# ============================================================
# GRAPH 02
# STATES
# ============================================================

create_bar_chart(
    df["State Name"],
    "Top 15 States by Number of Accidents",
    "State",
    "Number of Accidents",
    "02_top_states.png",
    rotation=45,
    limit=15
)


# ============================================================
# GRAPH 03
# CITIES
# ============================================================

create_bar_chart(
    df["City Name"],
    "Top 15 Cities by Number of Accidents",
    "City",
    "Number of Accidents",
    "03_top_cities.png",
    rotation=45,
    limit=15
)


# ============================================================
# GRAPH 04
# YEAR
# ============================================================

year_counts = (
    pd.to_numeric(
        df["Year"],
        errors="coerce"
    )
    .dropna()
    .astype(int)
    .value_counts()
    .sort_index()
)

plt.figure(
    figsize=(10, 5)
)

year_counts.plot(
    kind="bar"
)

plt.title(
    "Accidents by Year"
)

plt.xlabel(
    "Year"
)

plt.ylabel(
    "Number of Accidents"
)

plt.xticks(
    rotation=45
)

save_graph(
    "04_accidents_by_year.png"
)


# ============================================================
# GRAPH 05
# MONTH
# ============================================================

month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

month_counts = (
    df["Month"]
    .fillna("Unknown")
    .astype(str)
    .str.strip()
    .str.title()
    .value_counts()
)

# If month is numeric, convert numeric months
numeric_months = pd.to_numeric(
    df["Month"],
    errors="coerce"
)

if numeric_months.notna().sum() > 0:

    numeric_month_counts = (
        numeric_months
        .dropna()
        .astype(int)
        .value_counts()
        .sort_index()
    )

    numeric_month_counts.index = [
        month_order[i - 1]
        if 1 <= i <= 12
        else str(i)
        for i in numeric_month_counts.index
    ]

    month_counts = numeric_month_counts

else:

    month_counts = month_counts.reindex(
        month_order
    ).fillna(0)


plt.figure(
    figsize=(11, 5)
)

month_counts.plot(
    kind="bar"
)

plt.title(
    "Accidents by Month"
)

plt.xlabel(
    "Month"
)

plt.ylabel(
    "Number of Accidents"
)

plt.xticks(
    rotation=45,
    ha="right"
)

save_graph(
    "05_accidents_by_month.png"
)


# ============================================================
# GRAPH 06
# DAY OF WEEK
# ============================================================

day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_counts = (
    df["Day of Week"]
    .fillna("Unknown")
    .astype(str)
    .str.strip()
    .str.title()
    .value_counts()
)

day_counts = day_counts.reindex(
    day_order
).fillna(0)

plt.figure(
    figsize=(10, 5)
)

day_counts.plot(
    kind="bar"
)

plt.title(
    "Accidents by Day of Week"
)

plt.xlabel(
    "Day of Week"
)

plt.ylabel(
    "Number of Accidents"
)

plt.xticks(
    rotation=0
)

save_graph(
    "06_accidents_by_day_of_week.png"
)


# ============================================================
# GRAPH 07
# HOUR OF DAY
# ============================================================

time_values = (
    df["Time of Day"]
    .astype(str)
    .str.strip()
)

parsed_time = pd.to_datetime(
    time_values,
    format="%H:%M",
    errors="coerce"
)

# Try another time parser if needed
if parsed_time.notna().sum() == 0:

    parsed_time = pd.to_datetime(
        time_values,
        errors="coerce"
    )

hours = (
    parsed_time
    .dt.hour
    .dropna()
)

if len(hours) > 0:

    hour_counts = (
        hours
        .astype(int)
        .value_counts()
        .sort_index()
    )

    plt.figure(
        figsize=(11, 5)
    )

    plt.plot(
        hour_counts.index,
        hour_counts.values,
        marker="o"
    )

    plt.title(
        "Accidents by Hour of Day"
    )

    plt.xlabel(
        "Hour of Day"
    )

    plt.ylabel(
        "Number of Accidents"
    )

    plt.xticks(
        range(24)
    )

    save_graph(
        "07_accidents_by_hour.png"
    )


# ============================================================
# GRAPH 08
# VEHICLE TYPE
# ============================================================

create_bar_chart(
    df["Vehicle Type Involved"],
    "Accidents by Vehicle Type",
    "Vehicle Type",
    "Number of Accidents",
    "08_vehicle_type_distribution.png",
    rotation=40
)


# ============================================================
# GRAPH 09
# NUMBER OF VEHICLES
# ============================================================

vehicles = pd.to_numeric(
    df["Number of Vehicles Involved"],
    errors="coerce"
).dropna()

plt.figure(
    figsize=(9, 5)
)

vehicles.value_counts().sort_index().plot(
    kind="bar"
)

plt.title(
    "Number of Vehicles Involved"
)

plt.xlabel(
    "Number of Vehicles"
)

plt.ylabel(
    "Number of Accidents"
)

plt.xticks(
    rotation=0
)

save_graph(
    "09_number_of_vehicles.png"
)


# ============================================================
# GRAPH 10
# CASUALTIES
# ============================================================

casualties = pd.to_numeric(
    df["Number of Casualties"],
    errors="coerce"
).dropna()

plt.figure(
    figsize=(9, 5)
)

plt.hist(
    casualties,
    bins=15
)

plt.title(
    "Distribution of Accident Casualties"
)

plt.xlabel(
    "Number of Casualties"
)

plt.ylabel(
    "Number of Accidents"
)

save_graph(
    "10_casualties_distribution.png"
)


# ============================================================
# GRAPH 11
# FATALITIES
# ============================================================

fatalities = pd.to_numeric(
    df["Number of Fatalities"],
    errors="coerce"
).dropna()

plt.figure(
    figsize=(9, 5)
)

plt.hist(
    fatalities,
    bins=15
)

plt.title(
    "Distribution of Accident Fatalities"
)

plt.xlabel(
    "Number of Fatalities"
)

plt.ylabel(
    "Number of Accidents"
)

save_graph(
    "11_fatalities_distribution.png"
)


# ============================================================
# GRAPH 12
# WEATHER
# ============================================================

create_bar_chart(
    df["Weather Conditions"],
    "Accidents by Weather Conditions",
    "Weather Condition",
    "Number of Accidents",
    "12_weather_conditions.png",
    rotation=40
)


# ============================================================
# GRAPH 13
# ROAD TYPE
# ============================================================

create_bar_chart(
    df["Road Type"],
    "Accidents by Road Type",
    "Road Type",
    "Number of Accidents",
    "13_road_type.png",
    rotation=35
)


# ============================================================
# GRAPH 14
# ROAD CONDITION
# ============================================================

create_bar_chart(
    df["Road Condition"],
    "Accidents by Road Condition",
    "Road Condition",
    "Number of Accidents",
    "14_road_condition.png",
    rotation=0
)


# ============================================================
# GRAPH 15
# LIGHTING
# ============================================================

create_bar_chart(
    df["Lighting Conditions"],
    "Accidents by Lighting Conditions",
    "Lighting Condition",
    "Number of Accidents",
    "15_lighting_conditions.png",
    rotation=30
)


# ============================================================
# GRAPH 16
# TRAFFIC CONTROL
# ============================================================

create_bar_chart(
    df["Traffic Control Presence"],
    "Accidents by Traffic Control Presence",
    "Traffic Control",
    "Number of Accidents",
    "16_traffic_control.png",
    rotation=30
)


# ============================================================
# GRAPH 17
# SPEED LIMIT
# ============================================================

speed = pd.to_numeric(
    df["Speed Limit (km/h)"],
    errors="coerce"
).dropna()

plt.figure(
    figsize=(9, 5)
)

plt.hist(
    speed,
    bins=15
)

plt.title(
    "Speed Limit Distribution"
)

plt.xlabel(
    "Speed Limit (km/h)"
)

plt.ylabel(
    "Number of Accidents"
)

save_graph(
    "17_speed_limit_distribution.png"
)


# ============================================================
# GRAPH 18
# DRIVER AGE
# ============================================================

driver_age = pd.to_numeric(
    df["Driver Age"],
    errors="coerce"
).dropna()

plt.figure(
    figsize=(9, 5)
)

plt.hist(
    driver_age,
    bins=15
)

plt.title(
    "Driver Age Distribution"
)

plt.xlabel(
    "Driver Age"
)

plt.ylabel(
    "Number of Drivers"
)

save_graph(
    "18_driver_age_distribution.png"
)


# ============================================================
# GRAPH 19
# DRIVER GENDER
# ============================================================

create_bar_chart(
    df["Driver Gender"],
    "Accidents by Driver Gender",
    "Driver Gender",
    "Number of Accidents",
    "19_driver_gender.png",
    rotation=0
)


# ============================================================
# GRAPH 20
# LICENSE STATUS
# ============================================================

create_bar_chart(
    df["Driver License Status"],
    "Accidents by Driver License Status",
    "License Status",
    "Number of Accidents",
    "20_driver_license_status.png",
    rotation=25
)


# ============================================================
# GRAPH 21
# ALCOHOL
# ============================================================

create_bar_chart(
    df["Alcohol Involvement"],
    "Accidents by Alcohol Involvement",
    "Alcohol Involvement",
    "Number of Accidents",
    "21_alcohol_involvement.png",
    rotation=0
)


# ============================================================
# GRAPH 22
# LOCATION DETAILS
# ============================================================

create_bar_chart(
    df["Accident Location Details"],
    "Accidents by Location Details",
    "Location Details",
    "Number of Accidents",
    "22_location_details.png",
    rotation=30,
    limit=15
)


# ============================================================
# GRAPH 23
# SEVERITY BY WEATHER
# ============================================================

create_severity_chart(
    "Weather Conditions",
    "Accident Severity by Weather Conditions",
    "Weather Condition",
    "23_severity_by_weather.png"
)


# ============================================================
# GRAPH 24
# SEVERITY BY ROAD CONDITION
# ============================================================

create_severity_chart(
    "Road Condition",
    "Accident Severity by Road Condition",
    "Road Condition",
    "24_severity_by_road_condition.png"
)


# ============================================================
# GRAPH 25
# SEVERITY BY VEHICLE TYPE
# ============================================================

create_severity_chart(
    "Vehicle Type Involved",
    "Accident Severity by Vehicle Type",
    "Vehicle Type",
    "25_severity_by_vehicle_type.png"
)


# ============================================================
# GRAPH 26
# SEVERITY BY LIGHTING
# ============================================================

create_severity_chart(
    "Lighting Conditions",
    "Accident Severity by Lighting Conditions",
    "Lighting Condition",
    "26_severity_by_lighting.png"
)


# ============================================================
# GRAPH 27
# SEVERITY BY ALCOHOL
# ============================================================

create_severity_chart(
    "Alcohol Involvement",
    "Accident Severity by Alcohol Involvement",
    "Alcohol Involvement",
    "27_severity_by_alcohol.png",
    rotation=0
)


# ============================================================
# GRAPH 28
# SEVERITY BY YEAR
# ============================================================

severity_year = pd.crosstab(
    df["Year"],
    df["Accident Severity"]
)

plt.figure(
    figsize=(10, 6)
)

severity_year.plot(
    kind="bar",
    ax=plt.gca()
)

plt.title(
    "Accident Severity by Year"
)

plt.xlabel(
    "Year"
)

plt.ylabel(
    "Number of Accidents"
)

plt.xticks(
    rotation=45
)

plt.legend(
    title="Severity"
)

save_graph(
    "28_severity_by_year.png"
)


# ============================================================
# GRAPH 29
# SEVERITY BY MONTH
# ============================================================

severity_month = pd.crosstab(
    df["Month"],
    df["Accident Severity"]
)

# Convert numeric month indexes if needed
month_index_map = {
    1: "January",
    2: "February",
    3: "March",
    4: "April",
    5: "May",
    6: "June",
    7: "July",
    8: "August",
    9: "September",
    10: "October",
    11: "November",
    12: "December"
}

new_index = []

for value in severity_month.index:

    try:

        numeric_value = int(float(value))

        if numeric_value in month_index_map:
            new_index.append(
                month_index_map[numeric_value]
            )
        else:
            new_index.append(
                str(value)
            )

    except:

        new_index.append(
            str(value).strip().title()
        )

severity_month.index = new_index

ordered_rows = [
    month
    for month in month_order
    if month in severity_month.index
]

severity_month = severity_month.reindex(
    ordered_rows
)

plt.figure(
    figsize=(11, 6)
)

severity_month.plot(
    kind="bar",
    ax=plt.gca()
)

plt.title(
    "Accident Severity by Month"
)

plt.xlabel(
    "Month"
)

plt.ylabel(
    "Number of Accidents"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.legend(
    title="Severity"
)

save_graph(
    "29_severity_by_month.png"
)


# ============================================================
# GRAPH 30
# SEVERITY BY DAY OF WEEK
# ============================================================

severity_day = pd.crosstab(
    df["Day of Week"],
    df["Accident Severity"]
)

new_day_index = [
    str(value).strip().title()
    for value in severity_day.index
]

severity_day.index = new_day_index

ordered_days = [
    day
    for day in day_order
    if day in severity_day.index
]

severity_day = severity_day.reindex(
    ordered_days
)

plt.figure(
    figsize=(10, 6)
)

severity_day.plot(
    kind="bar",
    ax=plt.gca()
)

plt.title(
    "Accident Severity by Day of Week"
)

plt.xlabel(
    "Day of Week"
)

plt.ylabel(
    "Number of Accidents"
)

plt.xticks(
    rotation=0
)

plt.legend(
    title="Severity"
)

save_graph(
    "30_severity_by_day_of_week.png"
)


# ============================================================
# GRAPH 31
# SEVERITY BY ROAD TYPE
# ============================================================

create_severity_chart(
    "Road Type",
    "Accident Severity by Road Type",
    "Road Type",
    "31_severity_by_road_type.png",
    rotation=35
)


# ============================================================
# GRAPH 32
# SEVERITY BY DRIVER GENDER
# ============================================================

create_severity_chart(
    "Driver Gender",
    "Accident Severity by Driver Gender",
    "Driver Gender",
    "32_severity_by_driver_gender.png",
    rotation=0
)


# ============================================================
# GRAPH 33
# SEVERITY BY LICENSE STATUS
# ============================================================

create_severity_chart(
    "Driver License Status",
    "Accident Severity by Driver License Status",
    "License Status",
    "33_severity_by_license_status.png",
    rotation=25
)


# ============================================================
# GRAPH 34
# CASUALTIES BY SEVERITY
# ============================================================

plt.figure(
    figsize=(10, 6)
)

for severity in severity_counts.index:

    if severity == "Missing":
        continue

    values = pd.to_numeric(
        df.loc[
            df["Accident Severity"] == severity,
            "Number of Casualties"
        ],
        errors="coerce"
    ).dropna()

    if len(values) > 0:

        plt.hist(
            values,
            bins=15,
            alpha=0.5,
            label=str(severity)
        )

plt.title(
    "Casualties Distribution by Accident Severity"
)

plt.xlabel(
    "Number of Casualties"
)

plt.ylabel(
    "Number of Accidents"
)

plt.legend(
    title="Severity"
)

save_graph(
    "34_casualties_by_severity.png"
)


# ============================================================
# GRAPH 35
# FATALITIES BY SEVERITY
# ============================================================

plt.figure(
    figsize=(10, 6)
)

for severity in severity_counts.index:

    if severity == "Missing":
        continue

    values = pd.to_numeric(
        df.loc[
            df["Accident Severity"] == severity,
            "Number of Fatalities"
        ],
        errors="coerce"
    ).dropna()

    if len(values) > 0:

        plt.hist(
            values,
            bins=15,
            alpha=0.5,
            label=str(severity)
        )

plt.title(
    "Fatalities Distribution by Accident Severity"
)

plt.xlabel(
    "Number of Fatalities"
)

plt.ylabel(
    "Number of Accidents"
)

plt.legend(
    title="Severity"
)

save_graph(
    "35_fatalities_by_severity.png"
)


# ============================================================
# GRAPH 36
# CORRELATION HEATMAP
# ============================================================

numeric_for_corr = df.select_dtypes(
    include=np.number
)

correlation = (
    numeric_for_corr
    .corr()
)

correlation.to_csv(
    os.path.join(
        SUMMARY_DIR,
        "correlation_matrix.csv"
    )
)

if correlation.shape[1] >= 2:

    plt.figure(
        figsize=(12, 9)
    )

    plt.imshow(
        correlation,
        aspect="auto"
    )

    plt.colorbar(
        label="Correlation"
    )

    plt.xticks(
        range(
            len(correlation.columns)
        ),
        correlation.columns,
        rotation=90
    )

    plt.yticks(
        range(
            len(correlation.columns)
        ),
        correlation.columns
    )

    plt.title(
        "Numeric Feature Correlation Heatmap"
    )

    save_graph(
        "36_correlation_heatmap.png"
    )


# ============================================================
# 37. OUTLIER ANALYSIS
# ============================================================

outlier_rows = []

for column in numeric_columns:

    values = pd.to_numeric(
        df[column],
        errors="coerce"
    ).dropna()

    if len(values) < 4:
        continue

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = values[
        (values < lower_bound)
        | (values > upper_bound)
    ]

    outlier_rows.append({
        "Column": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier Percentage": (
            len(outliers)
            / len(values)
            * 100
        )
    })

if outlier_rows:

    outlier_summary = pd.DataFrame(
        outlier_rows
    )

    outlier_summary.to_csv(
        os.path.join(
            SUMMARY_DIR,
            "outlier_summary.csv"
        ),
        index=False
    )


# ============================================================
# 38. EDA REPORT
# ============================================================

report_lines = []

report_lines.append(
    "ACCIDENT HOTSPOT PREDICTION SYSTEM"
)

report_lines.append(
    "EXPLORATORY DATA ANALYSIS REPORT"
)

report_lines.append(
    "=" * 60
)

report_lines.append("")

report_lines.append(
    f"Dataset: {uploaded_csv}"
)

report_lines.append(
    f"Rows: {len(df)}"
)

report_lines.append(
    f"Columns: {len(df.columns)}"
)

report_lines.append(
    "Target: Accident Severity"
)

report_lines.append(
    f"Duplicate Rows: {duplicate_count}"
)

report_lines.append(
    f"Total Missing Values: {missing_total}"
)

report_lines.append("")

report_lines.append(
    "SEVERITY DISTRIBUTION"
)

report_lines.append(
    "-" * 40
)

for severity, count in severity_counts.items():

    percentage = (
        count
        / severity_counts.sum()
        * 100
    )

    report_lines.append(
        f"{severity}: "
        f"{count} "
        f"({percentage:.2f}%)"
    )

report_lines.append("")

report_lines.append(
    "GRAPH FILES GENERATED"
)

report_lines.append(
    "-" * 40
)

graph_files = sorted([
    filename
    for filename in os.listdir(
        GRAPHS_DIR
    )
    if filename.lower().endswith(".png")
])

for filename in graph_files:

    report_lines.append(
        filename
    )

with open(
    os.path.join(
        REPORTS_DIR,
        "eda_report.txt"
    ),
    "w",
    encoding="utf-8"
) as file:

    file.write(
        "\n".join(
            report_lines
        )
    )


# ============================================================
# 39. ZIP EDA OUTPUT
# ============================================================

zip_path = shutil.make_archive(
    "/content/accident_eda",
    "zip",
    EDA_DIR
)


# ============================================================
# 40. FINAL SUMMARY
# ============================================================

print("\n============================================================")
print("COMPLETE EDA FINISHED SUCCESSFULLY")
print("============================================================")

print(
    "\nDataset rows:",
    len(df)
)

print(
    "Dataset columns:",
    len(df.columns)
)

print(
    "Graphs saved:",
    len(graph_files)
)

print(
    "\nEDA folder:"
)

print(
    EDA_DIR
)

print(
    "\nZIP file:"
)

print(
    zip_path
)

print("\nGenerated graph files:")

for graph in graph_files:
    print(
        " -",
        graph
    )

print("\n============================================================")
print("All graphs were both SAVED and DISPLAYED in Colab.")
print("============================================================")
```
